```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt

import os

from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori

plt.style.use("ggplot")

plt.rcParams["figure.figsize"] = (10,6)

print("Libraries Loaded Successfully.")
```

```python
file_path = "/content/Dataset_B_Country_Features (2).csv"

df = pd.read_csv(file_path)

print("Dataset Loaded Successfully.")

display(df.head())
```

```python
output_folder = "/content/drive/MyDrive/COVID/AssociationRules"

tables_folder = os.path.join(output_folder,"Tables")
figures_folder = os.path.join(output_folder,"Figures")

os.makedirs(tables_folder,exist_ok=True)
os.makedirs(figures_folder,exist_ok=True)
```

```python
print("="*60)
print("DATASET INFORMATION")
print("="*60)

print("Shape")

print(df.shape)

print()

print("Columns")

print(df.columns.tolist())

print()

print("Missing Values")

print(df.isnull().sum())

print()

display(df.describe(include="all"))
```

```python
arm_df = df[[
    "Continent",
    "Origin",
    "First_Case_Month_Name",
    "First_Case_Quarter",
    "Death_Recorded",
    "Has_Travel_Origin",
    "Multiple_Origins",
    "Multiple_First_Day_Cases"
]].copy()

display(arm_df.head())
```

```python
transactions = []

for _, row in arm_df.iterrows():

    transaction = []

    for column in arm_df.columns:

        value = row[column]

        if pd.notna(value):

            transaction.append(f"{column}={value}")

    transactions.append(transaction)

print("Number of Transactions :",len(transactions))

print()

print("Example Transaction")

print(transactions[0])
```

```python
encoder = TransactionEncoder()

encoded = encoder.fit(transactions).transform(transactions)

transaction_df = pd.DataFrame(
    encoded,
    columns=encoder.columns_
)

display(transaction_df.head())
```

```python
transaction_df.to_csv(

    os.path.join(
        tables_folder,
        "TransactionDataset.csv"
    ),

    index=False
)

print("Transaction Dataset Saved.")
```

```python
frequent_items = apriori(

    transaction_df,

    min_support=0.05,

    use_colnames=True

)

frequent_items["Length"] = frequent_items["itemsets"].apply(len)

frequent_items = frequent_items.sort_values(

    by="support",

    ascending=False

)

display(frequent_items.head(20))
```

```python
frequent_items.to_csv(

    os.path.join(
        tables_folder,
        "FrequentItemsets.csv"
    ),

    index=False

)

print("Frequent Itemsets Saved.")
```

```python
length_count = (

    frequent_items["Length"]

    .value_counts()

    .sort_index()

)

plt.figure(figsize=(8,5))

bars = plt.bar(

    length_count.index.astype(str),

    length_count.values

)

plt.title(

    "Distribution of Frequent Itemset Lengths",

    fontsize=15,

    weight="bold"

)

plt.xlabel("Itemset Size")

plt.ylabel("Number of Itemsets")

for bar in bars:

    plt.text(

        bar.get_x()+bar.get_width()/2,

        bar.get_height(),

        int(bar.get_height()),

        ha="center",

        va="bottom"

    )

plt.tight_layout()

plt.savefig(

    os.path.join(

        figures_folder,

        "ItemsetLengthDistribution.pdf"

    ),

    dpi=300

)

plt.savefig(

    os.path.join(

        figures_folder,

        "ItemsetLengthDistribution.png"

    ),

    dpi=300

)

plt.show()
```

```python
top20 = frequent_items.head(20)

display(top20)

top20.to_csv(

    os.path.join(

        tables_folder,

        "Top20FrequentItemsets.csv"

    ),

    index=False

)
```

```python
from mlxtend.frequent_patterns import association_rules
```

```python
rules = association_rules(
    frequent_items,
    metric="lift",
    min_threshold=1.0
)

print("Total Rules Generated:", len(rules))

display(rules.head())
```

```python
rules = rules[[
    "antecedents",
    "consequents",
    "support",
    "confidence",
    "lift",
    "leverage",
    "conviction"
]]

rules["Antecedent Length"] = rules["antecedents"].apply(len)
rules["Consequent Length"] = rules["consequents"].apply(len)

display(rules.head())
```

```python
rules["Antecedents"] = rules["antecedents"].apply(
    lambda x: ", ".join(list(x))
)

rules["Consequents"] = rules["consequents"].apply(
    lambda x: ", ".join(list(x))
)

rules = rules.drop(columns=["antecedents","consequents"])

display(rules.head())
```

```python
rules.to_csv(
    os.path.join(
        tables_folder,
        "AssociationRules.csv"
    ),
    index=False
)

print("Association Rules Saved.")
```

```python
strong_rules = rules[
    (rules["support"] >= 0.05) &
    (rules["confidence"] >= 0.80) &
    (rules["lift"] >= 1.50)
].copy()

print("Strong Rules:", len(strong_rules))

display(strong_rules.head(20))
```

```python
strong_rules.to_csv(
    os.path.join(
        tables_folder,
        "StrongAssociationRules.csv"
    ),
    index=False
)
```

```python
top_lift = strong_rules.sort_values(
    by="lift",
    ascending=False
).head(20)

display(top_lift)

top_lift.to_csv(
    os.path.join(
        tables_folder,
        "Top20_Lift.csv"
    ),
    index=False
)
```

```python
top_confidence = strong_rules.sort_values(
    by="confidence",
    ascending=False
).head(20)

display(top_confidence)

top_confidence.to_csv(
    os.path.join(
        tables_folder,
        "Top20_Confidence.csv"
    ),
    index=False
)
```

```python
top_support = strong_rules.sort_values(
    by="support",
    ascending=False
).head(20)

display(top_support)

top_support.to_csv(
    os.path.join(
        tables_folder,
        "Top20_Support.csv"
    ),
    index=False
)
```
